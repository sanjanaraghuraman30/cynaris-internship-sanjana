```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Apply a professional visual theme
sns.set_theme(style="whitegrid", palette="deep")

# Create a folder to save our charts
output_dir = Path("charts")
output_dir.mkdir(exist_ok=True)

print("Libraries imported successfully!")
print(f"Charts will be saved in: {output_dir.resolve()}")
```

```python
import pandas as pd

# Create a sample sales dataset
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 13500, 18000, 22000, 25000],
    "Orders": [120, 145, 132, 175, 210, 240],
    "Marketing_Spend": [2000, 2500, 2200, 3000, 3500, 4000]
}

df = pd.DataFrame(data)

# Display the dataset
display(df)
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    df["Month"],
    df["Sales"],
    marker="o",
    linewidth=2,
    label="Monthly Sales"
)

plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.annotate(
    "Highest sales",
    xy=("Jun", 25000),
    xytext=("Apr", 24000),
    arrowprops={"arrowstyle": "->"}
)

plt.legend()
plt.tight_layout()

plt.savefig(output_dir / "line_chart.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python

# Create a line chart to show monthly sales trends
plt.figure(figsize=(9, 5))

plt.plot(
    df["Month"],
    df["Sales"],
    marker="o",
    linewidth=2,
    label="Monthly Sales"
)

# Add title and axis labels
plt.title("Monthly Sales Trend", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Sales")

# Highlight the highest sales value
plt.annotate(
    "Highest sales",
    xy=(5, 25000),
    xytext=(3, 24000),
    arrowprops={"arrowstyle": "->"}
)

plt.legend()
plt.tight_layout()

# Save the chart as a high-resolution PNG
plt.savefig(output_dir / "line_chart.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python

# Create a bar chart to compare monthly sales
plt.figure(figsize=(9, 5))

plt.bar(
    df["Month"],
    df["Sales"],
    label="Monthly Sales"
)

plt.title("Monthly Sales Comparison", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Sales")

# Add a value label above each bar
for i, sales in enumerate(df["Sales"]):
    plt.text(i, sales + 300, f"{sales:,}", ha="center")

plt.legend()
plt.tight_layout()

# Save the chart
plt.savefig(output_dir / "bar_chart.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python

# Create a scatter plot of marketing spend vs sales
plt.figure(figsize=(9, 5))

plt.scatter(
    df["Marketing_Spend"],
    df["Sales"],
    s=100,
    alpha=0.8,
    label="Monthly Data"
)

plt.title("Marketing Spend vs Sales", fontsize=16)
plt.xlabel("Marketing Spend")
plt.ylabel("Sales")

# Annotate each point with its month
for _, row in df.iterrows():
    plt.annotate(
        row["Month"],
        (row["Marketing_Spend"], row["Sales"]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.legend()
plt.tight_layout()

# Save the chart
plt.savefig(
    output_dir / "scatter_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# Create a histogram to show the distribution of sales
plt.figure(figsize=(9, 5))

plt.hist(
    df["Sales"],
    bins=5,
    edgecolor="black"
)

plt.title("Distribution of Monthly Sales", fontsize=16)
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.tight_layout()

# Save the histogram
plt.savefig(
    output_dir / "histogram.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# Create a box plot to examine sales distribution
plt.figure(figsize=(8, 5))

sns.boxplot(
    y=df["Sales"],
    width=0.35
)

plt.title("Distribution of Monthly Sales", fontsize=16)
plt.ylabel("Sales")

plt.tight_layout()

# Save the box plot
plt.savefig(
    output_dir / "box_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Explore sales data using Matplotlib and Seaborn to understand trends, comparisons, relationships, distributions, and variability.

- Month

## Visualisations
1. **Line Chart:** Shows monthly sales trends.

Different chart types answer different analytical questions. Clear titles, axis labels, annotations, and appropriate chart selection improve communication.
