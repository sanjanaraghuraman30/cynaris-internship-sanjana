## 2. Vectorised Operations and Aggregation

Vectorisation allows NumPy to perform operations on entire arrays without explicitly writing Python loops. This often improves performance for numerical workloads.

We will perform arithmetic operations and calculate aggregates across rows and columns using the axis parameter.

# Vectorised arithmetic
sales = np.array([1000, 1500, 2000, 2500, 3000])

print("Original sales:", sales)
print("Sales + 500:", sales + 500)

print("Sales with 10% increase:", sales * 1.10)

```python
# Vectorised arithmetic
sales = np.array([1000, 1500, 2000, 2500, 3000])

print("Original sales:", sales)
print("Sales + 500:", sales + 500)
print("Sales multiplied by 2:", sales * 2)
print("Sales with 10% increase:", sales * 1.10)

# Aggregation across axes
sales_matrix = np.array([
    [1000, 1500, 2000],
    [1200, 1800, 2200],
    [900, 1600, 2400]
])

print("\nSales matrix:\n", sales_matrix)
print("Total sales:", np.sum(sales_matrix))
print("Column totals (axis=0):", np.sum(sales_matrix, axis=0))
print("Row totals (axis=1):", np.sum(sales_matrix, axis=1))
print("Average sales:", np.mean(sales_matrix))
```

## 3. Broadcasting and Data Selection

Broadcasting allows NumPy to perform operations between compatible arrays of different shapes.

Boolean masking selects values that satisfy a condition, while fancy indexing selects specific positions using an index array.

```python
# Broadcasting
monthly_sales = np.array([
    [1000, 1200, 1400],
    [1500, 1700, 1900],
    [2000, 2200, 2400]
])

monthly_targets = np.array([100, 200, 300])

print("Monthly sales:\n", monthly_sales)
print("\nTargets:", monthly_targets)
print("\nSales after adding monthly targets:\n",
      monthly_sales + monthly_targets)

# Boolean masking
high_sales = monthly_sales[monthly_sales > 1800]
print("\nSales greater than 1800:", high_sales)

# Fancy indexing
selected_sales = monthly_sales[[0, 2], :]
print("\nFirst and third rows:\n", selected_sales)

# Select specific positions
specific_values = monthly_sales[[0, 1], [1, 2]]
print("\nSelected individual values:", specific_values)
```

## 4. Statistical Analysis and Correlation

NumPy provides functions to calculate descriptive statistics, including mean, standard deviation, percentiles, and correlation. These measures help analysts understand distributions, variability, and relationships between numerical variables.

```python
# Sample business dataset
revenue = np.array([12000, 15000, 18000, 22000, 25000, 30000])
orders = np.array([120, 145, 160, 190, 210, 250])

print("REVENUE STATISTICS")
print("Mean:", np.mean(revenue))
print("Median:", np.median(revenue))
print("Standard deviation:", np.std(revenue))
print("25th percentile:", np.percentile(revenue, 25))
print("75th percentile:", np.percentile(revenue, 75))
print("Minimum:", np.min(revenue))
print("Maximum:", np.max(revenue))

# Correlation between revenue and orders
correlation_matrix = np.corrcoef(revenue, orders)

print("\nCorrelation matrix:\n", correlation_matrix)
print("Revenue-orders correlation:", correlation_matrix[0, 1])
```

## 5. Performance Benchmark: Python vs NumPy

This benchmark compares a Python loop with NumPy vectorisation when multiplying one million values by two. Both methods perform the same calculation. Execution times depend on the computer and environment, so we measure them rather than assuming a fixed speedup.

```python
# Create one million values
n = 1_000_000
python_data = list(range(n))
numpy_data = np.arange(n)

# Python loop benchmark
start = time.perf_counter()

python_result = []
for value in python_data:
    python_result.append(value * 2)

python_time = time.perf_counter() - start

# NumPy vectorised benchmark
start = time.perf_counter()

numpy_result = numpy_data * 2

numpy_time = time.perf_counter() - start

print("PERFORMANCE BENCHMARK")
print(f"Dataset size: {n:,} values")
print(f"Python loop time: {python_time:.6f} seconds")
print(f"NumPy vectorised time: {numpy_time:.6f} seconds")

if numpy_time > 0:
    print(f"Speedup: {python_time / numpy_time:.2f}x")

# Verify both approaches produce the same result
print("Results match:", python_result == numpy_result.tolist())
```
