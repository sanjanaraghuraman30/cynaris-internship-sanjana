
# Week 2 — Day 2: Advanced Pandas Operations

## Learning Objectives
- Merge multiple DataFrames using different join types.
- Summarise data using pivot tables.
- Reshape data using `pd.melt()`.
- Combine operations using method chaining.

## Dataset
We will use sample sales and product datasets to practise advanced Pandas operations.


```python

import pandas as pd

# Sales dataset
sales = pd.DataFrame({
    "OrderID": [101, 102, 103, 104, 105],
    "Region": ["North", "South", "East", "West", "North"],
    "ProductID": [1, 2, 1, 3, 2],
    "Sales": [1200, 900, 1500, 700, 1100]
})

# Product dataset
products = pd.DataFrame({
    "ProductID": [1, 2, 3, 4],
    "Product": ["Laptop", "Phone", "Tablet", "Monitor"],
    "Category": ["Electronics", "Electronics", "Electronics", "Accessories"]
})

print("SALES DATA")
display(sales)

print("PRODUCT DATA")
display(products)
```

```python

# 1. INNER JOIN: only matching ProductIDs
inner_join = pd.merge(sales, products, on="ProductID", how="inner")
print("INNER JOIN")
display(inner_join)

# 2. LEFT JOIN: all sales rows, matching product details
left_join = pd.merge(sales, products, on="ProductID", how="left")
print("LEFT JOIN")
display(left_join)

# 3. RIGHT JOIN: all products, matching sales records
right_join = pd.merge(sales, products, on="ProductID", how="right")
print("RIGHT JOIN")
display(right_join)

# 4. OUTER JOIN: all sales and all products
outer_join = pd.merge(sales, products, on="ProductID", how="outer")
print("OUTER JOIN")
display(outer_join)
```

```python

# Create a pivot table: total sales by region and product
pivot_sales = pd.pivot_table(
    sales.merge(products, on="ProductID", how="left"),
    values="Sales",
    index="Region",
    columns="Product",
    aggfunc="sum",
    fill_value=0
)

print("TOTAL SALES BY REGION AND PRODUCT")
display(pivot_sales)
```

```python

# Create a wide-format dataset
monthly_sales = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet"],
    "January": [1200, 900, 700],
    "February": [1500, 1100, 850],
    "March": [1800, 1300, 950]
})

print("WIDE FORMAT")
display(monthly_sales)

# Convert wide format to long format
long_sales = pd.melt(
    monthly_sales,
    id_vars=["Product"],
    var_name="Month",
    value_name="Sales"
)

print("LONG FORMAT")
display(long_sales)
```

```python

# Filter, sort, and select columns in one pipeline
result = (
    sales
    .query("Sales >= 1000")
    .sort_values("Sales", ascending=False)
    .loc[:, ["OrderID", "Region", "Sales"]]
    .reset_index(drop=True)
)

print("FILTERED AND SORTED SALES")
display(result)
```


## Merge vs Concatenate

### `pd.merge()`
- Combines DataFrames using one or more common columns or keys.
- Supports inner, left, right, and outer joins.
- Example: Combining sales records with product details using `ProductID`.

### `pd.concat()`
- Combines DataFrames along rows or columns.
- Does not require a shared key column.
- Example: Appending January sales records to February sales records.

### Key Difference
`merge()` matches related records using keys, whereas `concat()` stacks or places DataFrames alongside each other.
